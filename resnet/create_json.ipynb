```python
import json
import random
from pathlib import Path


file_path = "/media/bmp10/da6a1340-5b8d-4bc2-9196-3a3bf678f27f/students/undergraduate/2022/WHF/discrep-mamaba/jsons/brats2023_gli.json"
files = Path(file_path)

jfile = json.loads(files.read_text(encoding="utf-8"))
ufile = jfile["train"] 

t1_samples, t1ce_samples, t2_samples, flair_samples = [], [], [], []

for usample in ufile:    
    images = usample["images"]
    t1_samples.append(images[0])   
    t1ce_samples.append(images[1]) 
    t2_samples.append(images[2]) 
    flair_samples.append(images[3]) 

train_samples = t1_samples[:250] + t1ce_samples[:250] + t2_samples[:250] + flair_samples[:250]
train_labels = [0]*250 + [1]*250 + [2]*250 + [3]*250

val_samples = t1_samples[250:] + t1ce_samples[250:] + t2_samples[250:] + flair_samples[250:]
val_labels = [0]*50 + [1]*50 + [2]*50 + [3]*50

train_data = [{"image": k, "label": v} for k, v in zip(train_samples, train_labels)]
val_data = [{"image": k, "label": v} for k, v in zip(val_samples, val_labels)]


random.shuffle(val_data)

# 6. 整合并保存到 JSON 文件
final_json = {
    "train": train_data,
    "val": val_data
}

output_path = "./brats_modality.json"
with open(output_path, "w", encoding="utf-8") as f:
    json.dump(final_json, f, indent=4, ensure_ascii=False)

print(f"数据处理完成！已保存至: {output_path}")
print(f"训练集数量: {len(train_data)}, 验证集数量: {len(val_data)}")
```

```python
import json
import random
from pathlib import Path


file_path = "/media/bmp10/da6a1340-5b8d-4bc2-9196-3a3bf678f27f/students/undergraduate/2022/WHF/discrep-mamaba/jsons/brats2023_gli.json"
files = Path(file_path)

jfile = json.loads(files.read_text(encoding="utf-8"))
ufile = jfile["val"] 
num = len(ufile)
t1_samples, t1ce_samples, t2_samples, flair_samples = [], [], [], []

for usample in ufile:    
    images = usample["images"]
    t1_samples.append(images[0])   
    t1ce_samples.append(images[1]) 
    t2_samples.append(images[2]) 
    flair_samples.append(images[3]) 

val_samples = t1_samples + t1ce_samples + t2_samples + flair_samples
val_labels = [0]*num + [1]*num + [2]*num + [3]*num

val_data = [{"image": k, "label": v} for k, v in zip(val_samples, val_labels)]


random.shuffle(val_data)

# 6. 整合并保存到 JSON 文件
final_json = {
    "val": val_data
}

output_path = "./brats_modality_test.json"
with open(output_path, "w", encoding="utf-8") as f:
    json.dump(final_json, f, indent=4, ensure_ascii=False)

print(f"数据处理完成！已保存至: {output_path}")
print(f"训练集数量: {len(train_data)}, 验证集数量: {len(val_data)}")
```
