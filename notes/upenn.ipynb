```python
import os 
import json
from pathlib import Path

from monai.transforms import LoadImage
loader = LoadImage()

seg_root = "/media/bmp10/da6a1340-5b8d-4bc2-9196-3a3bf678f27f/students/undergraduate/2022/WHF/datasets/upenn_gbm/tumor_seg"
t1_root = "/media/bmp10/da6a1340-5b8d-4bc2-9196-3a3bf678f27f/students/undergraduate/2022/WHF/datasets/upenn_gbm/t1w"
t1ce_root = "/media/bmp10/da6a1340-5b8d-4bc2-9196-3a3bf678f27f/students/undergraduate/2022/WHF/datasets/upenn_gbm/t1wgd"
t2_root = "/media/bmp10/da6a1340-5b8d-4bc2-9196-3a3bf678f27f/students/undergraduate/2022/WHF/datasets/upenn_gbm/t2w"
flair_root = "/media/bmp10/da6a1340-5b8d-4bc2-9196-3a3bf678f27f/students/undergraduate/2022/WHF/datasets/upenn_gbm/flair"
sample_list = os.listdir(seg_root)

sample_list = [f'{name.split("_")[0]}_{name.split("_")[1]}' for name in sample_list]

samples = []
for name in sample_list:
    sample = {}
    sample["images"] = [
        os.path.join(t1_root,f"{name}_T1.nii.gz"),
        os.path.join(t1ce_root, f"{name}_T1GD.nii.gz"),
        os.path.join(t2_root, f"{name}_T2.nii.gz"),
        os.path.join(flair_root, f"{name}_FLAIR.nii.gz")
    ]
    sample["seg"] = os.path.join(seg_root,f"{name}_segm.nii.gz")
    sample["label"] = -1
    samples.append(sample)

num = len(samples)
print(len(samples))
train_num = int(0.7*num)
data = {}
data["train"] = samples[:train_num]
data["val"] = samples[train_num:]



file_path = Path("/media/bmp10/da6a1340-5b8d-4bc2-9196-3a3bf678f27f/students/undergraduate/2022/WHF/discrep-mamaba/jsons/ucsf_pdgm.json")
file_path.write_text(
    json.dumps(data, ensure_ascii=False, indent=2),
    encoding='utf-8'
)

```
